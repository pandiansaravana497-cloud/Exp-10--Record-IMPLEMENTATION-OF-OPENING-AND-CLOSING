```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load the input image
image_path = r"C:\Users\acer\OneDrive\Pictures\Camera Roll\road.jpg"
img = cv2.imread(image_path)

# Check whether image is loaded
if img is None:
    print("Error: Image not found!")
else:
    # Convert BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Step 2: Display original image
    plt.figure(figsize=(5, 5))
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()

    # Step 3: Convert image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 4: Apply binary thresholding
    _, binary = cv2.threshold(
        gray, 127, 255, cv2.THRESH_BINARY
    )

    # Step 5: Create structuring element
    kernel = np.ones((5, 5), np.uint8)

    # Step 6: Opening operation (Erosion + Dilation)
    opening = cv2.morphologyEx(
        binary, cv2.MORPH_OPEN, kernel
    )

    # Step 7: Closing operation (Dilation + Erosion)
    closing = cv2.morphologyEx(
        binary, cv2.MORPH_CLOSE, kernel
    )

    # Step 8: Display all results
    titles = [
        "Original Image",
        "Binary Image",
        "Opening Operation",
        "Closing Operation"
    ]

    images = [
        img_rgb,
        binary,
        opening,
        closing
    ]

    plt.figure(figsize=(12, 8))

    for i in range(4):
        plt.subplot(2, 2, i + 1)
        if i == 0:
            plt.imshow(images[i])
        else:
            plt.imshow(images[i], cmap="gray")

        plt.title(titles[i])
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```
